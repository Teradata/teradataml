<b style = 'font-size:28px;font-family:Arial;color:#E37C4D'>Import libraries and create connection.</b>

```python
import getpass
from teradataml import create_context
from teradatasqlalchemy.types import ARRAY_INTEGER, ARRAY_VARCHAR

# Get the connection to the Vantage using create_context()
ctx = create_context(host=getpass.getpass("Hostname: "), username=getpass.getpass("Username: "), password=getpass.getpass("Password: "))

ctx.load_example_data("array", "array_table")
ctx.load_example_data("dataframe", "admissions_train")

# Create a DataFrame on 'array_table'.
df_arrays = ctx.DataFrame("array_table")
tdf = ctx.DataFrame("admissions_train")
```

```python
from IPython.display import display
from teradataml import display as td_display
td_display.enable_ui = False
def print_variables(df):
    print(" ************************* DataFrame ********************* ")
    display(df)
    print("\n\n")
    print(" ************************* DataFrame.tdtypes ********************* ")
    display(df.tdtypes)
    print("\n\n")
```

```python
print_variables(df_arrays)
```

```python
print_variables(tdf)
```

<hr>
<b style = 'font-size:28px;font-family:Arial;color:#E37C4D'>Array</b>

```python
# Create an Array object with integer values and assign it to a new column.
arr_obj = ctx.Array((1, 2, 3, 4, 5), ARRAY_INTEGER('[5]'))
df = df_arrays.assign(new_array = arr_obj)
print_variables(df)
```

```python
# Create an Array object with VARCHAR values.
arr_varchar = ctx.Array(('hello', 'world', 'test'), ARRAY_VARCHAR('[3]'))
df = df_arrays.assign(string_array = arr_varchar)
print_variables(df)
```

<b style = 'font-size:28px;font-family:Arial;color:#E37C4D'>DataFrame</b>

### DataFrame.unnest() Method

```python
# Explode the array column 'arr1' without key column and ordinality.
res = df_arrays.unnest(array_col="arr1")
print_variables(res)
```

```python
# Explode the array column 'arr3' with key column 'id' and ordinality.
df = df_arrays.unnest(array_col=df_arrays.arr3, key_col="id", ordinality=True)
print_variables(df)
```
